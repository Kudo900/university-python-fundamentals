```python
for rows in range(1, 6):
  start_num = rows * 2
  print(start_num, end=" ")
  for col in range(1, start_num+1):
    number = start_num - col
    print(number, end=" ")
  print()  # Add a newline after each row's countdown completes
```

```python
user = input("Enter username:")
user = user.lower()
password = ""

for i in range(0, len(user)):
  char = user[i]
  if char == "i":
    char = "1"
  elif char == "r":
    char = "7"
  elif char == "s":
    char = "5"
  elif char == "z":
    char = "2"
  password = password + char

print("Password is", password)
```
